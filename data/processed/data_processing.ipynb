```python
# Drop duplicates
df.drop_duplicates(subset='id', inplace=True)
```

```python
# Considering dropping or keeping series with no description or tags in a separate pool
no_tags_desc = df[(df['tags'].isnull() & df['description'].isnull())]
ids_no_tags_desc = no_tags_desc['id']

# Dropping ids with no tags or descriptions for model training
df.drop(df[df['id'].isin(ids_no_tags_desc)].index, inplace=True)
```
